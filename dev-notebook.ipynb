```python

new_collegues = ["Aleksei","Amine","Anna","Astha","Brigitta",
                 "Bryan","Ena","Esra","Faranges","Frédéric",
                 "Hamideh","Héloïse","Imran","Intan K.",
                 "Jens","Kristin","Michiel","Nancy","Pierrick",
                 "Sandrine","Tim","Viktor","Welederufeal","Živile"]

class Seat:

    """This class represents a seat and displays whether it is free or occupied."""

    def __init__(self, free: bool = True, occupant: str | None = None) -> None:
    
        """First, we initialize a Seat object with default attributes.

        :param free: A boolean indicating if the seat is free (True by default).
        :param occupant: A string representing the current occupant of the seat, or None if the seat is free.
        :return: None
        """
        self.free: bool = free
        self.occupant: str |None = occupant
        
    
    def set_occupant(self, name: str) -> None:

        """Second, if a seat is free, we wish to assign somebody to it.

        :param name: A string representing the name of the person to occupy the seat.
        :return: None
        :raises ValueError: If the seat is already occupied.
        """
        if self.free:
            # Here we assign an occupant and mark the seat as occupied.
            self.occupant = name
            self.free = False
            print(f"This seat is free, {name} - please take the seat and enjoy!")
        else:
            # This avoids overwriting an occupied seat.
            print("This seat is already occupied. Please check another one.")

    def remove_occupant(self, name):

        """ Third, when we remove a person forma seat, we wish to return their name.
        The return will be a string containing the name of the previous occupant, or None if the seat was already free."""
        
        if not self.free:
            # Here we store the name of the previous occupant before we remove them.
            previous_occupant = self.occupant
            self.occupant = None
            self.free = True
            print(f"This seat was just made available again by {previous_occupant}")
            return previous_occupant 
        else: 
            # Here, the seat is already free.
            print("This seat is free.")
            return None 


    
```

```python
class Table:

    """ This class represents a table with a set number of seats.
    It can assign seats and also track the capacity of free seats."""

    def __init__(self, capacity= int) -> None:

        """ First, we initialize a Table with a set number of seats.

        :param capacity: An integer representing the total number of seats at the table.
        :return: None
        """
        self.capacity: int = capacity  
               #   This gives us the total number of seats.
        self.seats = [Seat() for _ in range(capacity)]
               # This creates the seats.

    def has_free_spot(self = True) -> bool:

        """Second, we check if the table has a free seat.
        The return will be a boolean giving if a least one seat is free."""
    
        return any(seat.free for seat in self.seats)

    def assign_seat(self, name: str)-> bool:

        """Third, we assign a seat to someone if the seat is free.

        :param name: A string representing the name of the person to assign a seat to.
        :return: True if a seat was successfully assigned, False if the table is full.
        """
               
        for seat in self.seats: 
            if seat.free:
               seat.set_occupant(name)
               return True
        print(f"There are no free spaces for you, {name}. I'm sorry for that.")
        return False
        
    def left_capacity(self) -> int:

        """Fourth, we calculate the number of available free seats at the table.
        
        :return: An integer representing how many seats are still free."""
        return sum(seat.free for seat in self.seats)
    
    
    def __str__(self) -> str:

        """
        Last, here we create a string representation of the table's current status.

        :return: A string describing how many seats are occupied and how many are free.
        """
        occupied = self.capacity - self.left_capacity()
        return (
            f"Table capacity: {self.capacity} | "
            f"Occupied: {occupied} | "
            f"Free: {self.left_capacity()}")

    
```

```python
import random

class Openspace:

    """Here we create a class that represents the openspace with the tables, each with a specific number of seats."""

    def __init__(self, number_of_tables: int, table_capacity: int) -> None:

        """First, we initialize an Openspace with a number of tables and fixed number of seats per table.

        :param number_of_tables: The total number of tables in the openspace.
        :param table_capacity: The number of seats per table.
        :return: None. """

        self.number_of_tables: int = number_of_tables
        self.tables: list[Table] = [Table(table_capacity) for _ in range(number_of_tables)]

    def organize(self, names: list[str]) -> None:

        """Second, we randomly assign the people to the seats across the tables.

        :param names: A list of strings representing the names of people to assign to seats.
        :return: None"""
        random.shuffle(names)
            #  This will shuffle the names randomly when placing them on the seats.
        for name in names:
            assigned = False
            for table in self.tables:
                if table.has_free_spot():
                    table.assign_seat(name)
                    assigned = True
                    break
            if not assigned:
                # This means no seats are free.
                print(f"No more free seats for {name}.")

    def __str__(self) -> str:

        """Third, we show how the persons will be seated around the tables, readable to all.

        :return: None. """
        lines: list[str] = ["Openspace seat arrangement\n"]
        for i, table in enumerate(self.tables, start=1):
            lines.append(f"Table {i}:")
            for j, seat in enumerate(table.seats, start=1):
                occupant = seat.occupant if not seat.free else "Free"
                lines.append(f"  Seat {j}: {occupant}")
            lines.append("")  
                  # This provides us with a blank line between tables
        return "\n".join(lines)

    def display(self) -> None:
         
         """Fourth, we display the previous string of how people are seated in a format readable to anyone.
         
         :return: None."""
         print(self.__str__())

    def store (self, filename: str)-> None:
         
        """Last, we store the seat arrangement in a text file.
         
        :return: None."""
        with open(filename, "w", encoding="utf-8") as file:
            file.write(str(self))
        print(f"Seating arrangement successfully written to '{filename}'.")
         
    
        

        
```

```python
new_Openspace= Openspace (number_of_tables=6 , table_capacity=4)
names = ["Aleksei","Amine","Anna","Astha","Brigitta",
                 "Bryan","Ena","Esra","Faranges","Frédéric",
                 "Hamideh","Héloïse","Imran","Intan K.",
                 "Jens","Kristin","Michiel","Nancy","Pierrick",
                 "Sandrine","Tim","Viktor","Welederufeal","Živile"]
new_Openspace.organize(names)
new_Openspace.display()
new_Openspace.store("seating_plan.txt")
print("seating_plan.txt")
```
